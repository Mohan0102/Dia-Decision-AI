```python
import pandas as pd
import glob
import os

folder = "../data/raw/Output_CSV"

files = glob.glob(os.path.join(folder, "*.csv"))

print(files)

dfs = {}

for file in files:
    name = os.path.basename(file).replace(".csv","")
    dfs[name] = pd.read_csv(file)

print(dfs.keys())
```

```python
for name, df in dfs.items():
    print("="*60)
    print(name.upper())
    print("="*60)

    print("shape:", df.shape)
    print("\ncolumns:", df.columns.tolist())

    print("\n Data Type")
    print(df.dtypes)

    print("\nFirst 5 Rows")
    display(df.head())
```

```python
missing = {}

for name, df in dfs.items():
    missing[name] = df.isnull().sum()

for name, values in missing.items():
    print('\n',name)
    print(values)
```

```python
for name, df in dfs.items():
    print(name, ':', df.duplicated().sum())
```

```python
for name, df in dfs.items():
    print('='*60)
    print(name)
    print('='*60)

    print(df.describe(include='all'))
```

```python
for name, df in dfs.items():
    if 'ts' in df.columns:
        df['ts'] = pd.to_datetime(
            df['ts'],
            format = 'mixed',
            dayfirst = True,
            errors = 'coerce'
        )

    if 'ts_begin' in df.columns:
        df['ts_begin'] = pd.to_datetime(
            df['ts_begin'],
            format = 'mixed',
            dayfirst = True,
            errors = 'coerce'
        )

    if 'ts_end' in df.columns:
        df['ts_end'] = pd.to_datetime(
            df['ts_end'],
            format = 'mixed',
            dayfirst = True,
            errors = 'coerce'
        )
    
    if 'tbegin' in df.columns:
        df['tbegin'] = pd.to_datetime(
            df['tbegin'],
            format = 'mixed',
            dayfirst = True,
            errors = 'coerce'
        )
    
    if 'tend' in df.columns:
        df['tend'] = pd.to_datetime(
            df['tend'],
            format = 'mixed',
            dayfirst = True,
            errors = 'coerce'
        )
print("All timestamps converted successfully!")
```

```python
for name, df in dfs.items():

    print(f"\n{name}")

    for col in ["ts", "ts_begin", "ts_end", "tbegin", "tend"]:
        if col in df.columns:
            print(f"{col}: {df[col].isna().sum()} invalid values")
```

```python
for name, df in dfs.items():
    print(f'\n{name}')

    if 'ts' in df.columns:
        print('Start:', df['ts'].min())
        print('End:', df['ts'].max())

    elif 'ts_begin' in df.columns:
        print('Start:', df['ts_begin'].min())
        print('End:', df['ts_end'].max())

    elif 'tbegin' in df.columns:
        print('Start:', df['tbegin'].min())
        print('End:', df['tend'].max())
    
    print("Total Records:", len(df))
```

```python
for name, df in dfs.items():
    print(f"\n{name}")
    print("Unique Patients:", df["Patient"].nunique())
    print(df["Patient"].unique())
```

```python
for name, df in dfs.items():
    before = len(df)

    df.drop_duplicates(inplace=True)

    after = len(df)

    print(f"{name}: Removed {before-after} duplicate rows")
```

```python
for name, df in dfs.items():

    if "ts" in df.columns:
        df.sort_values("ts", inplace=True)

    elif "ts_begin" in df.columns:
        df.sort_values("ts_begin", inplace=True)

    elif "tbegin" in df.columns:
        df.sort_values("tbegin", inplace=True)
```

```python
for df in dfs.values():
    df.reset_index(drop=True, inplace=True)
```

```python
for name, df in dfs.items():

    print(name)

    print(df.info())

    print()
```

```python
import os

processed_folder = "../data/processed"
os.makedirs(processed_folder, exist_ok=True)

for name, df in dfs.items():
    df.to_csv(f"{processed_folder}/{name}.csv", index=False)

print("All cleaned files saved!")
```
